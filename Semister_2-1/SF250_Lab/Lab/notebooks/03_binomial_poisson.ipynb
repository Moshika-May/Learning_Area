# Binomial vs Poisson Approximation

When n is large and p is small, the Binomial distribution B(n, p) is very
close to the Poisson distribution Po(np). This program simulates a Binomial
experiment and puts the observed frequencies next to both exact
distributions so you can see how close they are.

## Build and run

```
make
./binomial_poisson [n] [p] [samples] [seed]
```

| Argument | Meaning | Default |
|---|---|---|
| `n` | trials per experiment (>= 1) | 100 |
| `p` | success probability per trial, 0 < p < 1 | 0.05 |
| `samples` | how many experiments to simulate (>= 1) | 100000 |
| `seed` | random seed | current time |

The run costs about `samples * n` random draws, so very large values of both
make it slow.

## How it works

1. **Simulation.** One experiment makes `n` Bernoulli(p) trials and counts the
   successes `x`. This repeats `samples` times, and the frequency of each
   `x` is recorded.
2. **Exact probabilities.** For each r it computes
   - Binomial: `C(n,r) p^r (1-p)^(n-r)`
   - Poisson with `lambda = np`: `lambda^r e^(-lambda) / r!`

   Both are computed in log form with `lgamma`, so large n does not overflow.
3. **Table.** Rows run from r = 0 up to about `lambda + 6*sqrt(lambda) + 5`
   (or n if smaller), where the probabilities are negligible. The `#` bars
   show the observed frequency.
4. **Summary.** It prints the observed mean and variance next to the theory
   (Binomial: `np` and `np(1-p)`, Poisson: both equal `lambda`), the sum of
   `|Binomial - Poisson|` over the shown rows, and whether the rule of thumb
   `n > 50` and `p < 0.1` holds.

## Makefile shortcuts

```
make                          build ./binomial_poisson
make run                      build and run: ./binomial_poisson 100 0.05 100000 1
make run ARGS="10 0.5 50000"  run with your own arguments

```

## 4. `binomial_poisson`: ประมาณ Binomial ด้วย Poisson

**คืออะไร:** เมื่อ n มากและ p น้อย Binomial B(n, p) จะใกล้เคียง Poisson Po(np) lab นี้จำลองแล้วเทียบ pmf ทั้งสองแบบ
รัน: `./binomial_poisson [n=100] [p=0.05] [samples=100000] [seed]`

**ทำงานยังไง:**
1. จำลองการทดลอง Bernoulli n ครั้ง นับจำนวนสำเร็จ ทำซ้ำ `samples` รอบ
2. พิมพ์ความถี่ที่ได้จริง เทียบกับ pmf ของ Binomial และ Poisson (λ = np) ที่แต่ละ r พร้อมแท่งกราฟ
3. เทียบ mean และ var กับทฤษฎี และรายงานผลต่างรวมของ pmf สองแบบ
4. ลอง `./binomial_poisson 100 0.05` (ใกล้กันมาก) เทียบกับ `./binomial_poisson 10 0.5` (ต่างกันชัด) เพื่อดูว่ากฎ n > 50, p < 0.1 มีความหมายอย่างไร

| ส่วนของโปรแกรม | อ้างอิงสไลด์ |
|---|---|
| Binomial: จำนวนสำเร็จใน n ครั้ง, pmf = nCr·p^r·(1−p)^(n−r) | DD 7 |
| nCr และปัญหา n! ใหญ่เกินไป | DD 8 |
| ตัวอย่าง CPU เสียบน wafer (n=100, p=0.05) ซึ่งเป็นค่าเริ่มต้นของ lab | DD 9 |
| กราฟ Binomial ที่พารามิเตอร์ต่างๆ | DD 10 |
| E[X] = np, Var = np(1−p) | DD 11-12 |
| ที่มาของ Poisson: แบ่งเวลาเป็นช่วงเล็กๆ แต่ละช่วงเป็น Bernoulli, p = λ/n, n → ∞ | DD 13-14 |
| Poisson pmf = λʳe^(−λ)/r!, E = Var = λ | DD 15 |
| กราฟ Poisson และการประมาณ B(n,p) ≈ Po(np) เมื่อ n > 50, p < 0.1 | DD 16 |

⚠ **นอกสไลด์:** การใช้ `lgamma` คำนวณแฟกทอเรียลใน log-space (สไลด์ DD 8 แนะนำ Stirling สำหรับปัญหาเดียวกัน) และตัวชี้วัดผลต่างรวม Σ|Binomial − Poisson|

## Source code

One cell per file. Run them all to recreate the project.

````python
%%writefile binomial_poisson/binomial_poisson.h
#ifndef BINOMIAL_POISSON_H
# define BINOMIAL_POISSON_H

# include <math.h>
# include <stdio.h>
# include <stdlib.h>
# include <time.h>

# define W 40

typedef struct s_bp
{
	long			n;
	long			samples;
	long			top;
	long			*h;
	double			p;
	double			lam;
	double			s1;
	double			s2;
	double			tv;
	unsigned int	seed;
}	t_bp;

double			lbin(long n, long r, double p);
double			lpoi(double l, long r);
void			simulate(t_bp *b);
void			print_table(t_bp *b);
void			print_stats(t_bp *b);

#endif
````

````python
%%writefile binomial_poisson/binomial_poisson.c
#include "binomial_poisson.h"

static int	parse(t_bp *b, int ac, char **av)
{
	b->n = 100;
	b->p = 0.05;
	b->samples = 100000;
	b->seed = (unsigned)time(NULL);
	if (ac > 1)
		b->n = atol(av[1]);
	if (ac > 2)
		b->p = atof(av[2]);
	if (ac > 3)
		b->samples = atol(av[3]);
	if (ac > 4)
		b->seed = (unsigned)atol(av[4]);
	if (b->n < 1 || b->samples < 1 || b->p <= 0 || b->p >= 1)
	{
		fprintf(stderr, "usage: %s [n>=1] [0<p<1] [samples>=1] [seed]\n",
			av[0]);
		return (1);
	}
	return (0);
}

int	main(int ac, char **av)
{
	t_bp	b;

	if (parse(&b, ac, av))
		return (1);
	srand(b.seed);
	b.lam = b.n * b.p;
	b.top = (long)ceil(b.lam + 6 * sqrt(b.lam) + 5);
	if (b.top > b.n)
		b.top = b.n;
	b.h = calloc(b.top + 1, sizeof(long));
	simulate(&b);
	printf("== Binomial B(n=%ld, p=%g) ", b.n, b.p);
	printf("vs Poisson Po(lambda=np=%g), ", b.lam);
	printf("%ld samples, seed=%u ==\n", b.samples, b.seed);
	print_table(&b);
	print_stats(&b);
	free(b.h);
	return (0);
}
````

````python
%%writefile binomial_poisson/binomial_poisson_math.c
#include "binomial_poisson.h"

double	lbin(long n, long r, double p)
{
	return (lgamma(n + 1.0) - lgamma(r + 1.0) - lgamma(n - r + 1.0) + r * log(p)
		+ (n - r) * log1p(-p));
}

double	lpoi(double l, long r)
{
	return (r * log(l) - l - lgamma(r + 1.0));
}

void	simulate(t_bp *b)
{
	long	i;
	long	j;
	long	x;

	i = -1;
	b->s1 = 0;
	b->s2 = 0;
	while (++i < b->samples)
	{
		x = 0;
		j = -1;
		while (++j < b->n)
			if (rand() / (RAND_MAX + 1.0) < b->p)
				x++;
		b->s1 += x;
		b->s2 += (double)x * x;
		if (x <= b->top)
			b->h[x]++;
	}
}
````

````python
%%writefile binomial_poisson/binomial_poisson_print.c
#include "binomial_poisson.h"

static long	max_count(t_bp *b)
{
	long	mx;
	long	r;

	mx = 0;
	r = -1;
	while (++r <= b->top)
		if (b->h[r] > mx)
			mx = b->h[r];
	return (mx);
}

static void	bar(long v, long mx)
{
	int	k;

	k = 0;
	while (k < (int)(W * (double)v / mx + 0.5))
	{
		putchar('#');
		k++;
	}
	putchar('\n');
}

void	print_table(t_bp *b)
{
	long	r;
	long	mx;
	double	bv;
	double	q;

	mx = max_count(b);
	b->tv = 0;
	printf("%3s %9s %9s %9s\n", "r", "observed", "Binomial", "Poisson");
	r = -1;
	while (++r <= b->top)
	{
		bv = exp(lbin(b->n, r, b->p));
		q = exp(lpoi(b->lam, r));
		b->tv += fabs(bv - q);
		printf("%3ld %9.5f %9.5f %9.5f ", r, (double)b->h[r] / b->samples, bv,
			q);
		bar(b->h[r], mx);
	}
}

void	print_stats(t_bp *b)
{
	double	m;
	double	v;

	m = b->s1 / b->samples;
	v = b->s2 / b->samples - m * m;
	printf("\n          observed   Binomial   Poisson\n");
	printf("mean   %10.4f %10.4f %10.4f\n", m, b->lam, b->lam);
	printf("var    %10.4f %10.4f %10.4f   (Poisson: mean = var)\n", v, b->lam
		* (1 - b->p), b->lam);
	printf("sum |Binomial - Poisson| over shown r = %.5f\n", b->tv);
	printf("Rule of thumb (n > 50, p < 0.1): ");
	if (b->n > 50 && b->p < 0.1)
		printf("satisfied, approximation should be good\n");
	else
		printf("NOT satisfied, expect a visible gap\n");
}
````

````python
%%writefile binomial_poisson/Makefile
NAME = binomial_poisson
SRCS = binomial_poisson.c binomial_poisson_math.c binomial_poisson_print.c
CFLAGS = -Wall -Wextra -O2
ARGS = 100 0.05 100000 1

all: $(NAME)

$(NAME): $(SRCS) binomial_poisson.h
	$(CC) $(CFLAGS) $(SRCS) -o $(NAME) -lm

# make run               runs the example above
# make run ARGS="..."    runs with your own arguments
run: $(NAME)
	./$(NAME) $(ARGS)

clean:
	rm -f $(NAME)

re: clean all

.PHONY: all run clean re
````
