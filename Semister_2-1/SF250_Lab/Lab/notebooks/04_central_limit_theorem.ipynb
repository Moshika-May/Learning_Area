# Central Limit Theorem Demo

The Central Limit Theorem says that the average of many independent draws
looks Normal, even when each draw is not Normal. This program draws samples
from a skewed or flat population, computes many sample means, and compares
their distribution with the standard Normal curve.

## Build and run

```
make
./central_limit_theorem [uniform|exp] [n] [reps] [seed]
```

| Argument | Meaning | Default |
|---|---|---|
| `uniform` or `exp` | population: Uniform(0,1) or Exponential(1) | `uniform` |
| `n` | sample size (draws per mean, >= 1) | 30 |
| `reps` | how many sample means to compute (>= 1) | 10000 |
| `seed` | random seed | current time |

## How it works

1. **Population.** Uniform(0,1) has mean 1/2 and variance 1/12. Exponential(1)
   has mean 1 and variance 1, and is strongly skewed to the right. Exponential
   values are made from uniform ones with `-ln(u)`.
2. **Sampling.** For each of `reps` repetitions it averages `n` draws to get
   a sample mean `xbar`.
3. **Standardising.** Each mean becomes `z = (xbar - mu) / (sigma / sqrt(n))`.
   By the CLT, z is close to N(0, 1) when n is large.
4. **Histogram.** The z values fall into 16 bins of width 0.5 covering
   [-4, 4). Next to each bin it prints the count expected from N(0, 1),
   computed with the Normal CDF. Values outside [-4, 4) are not shown.
5. **Summary.** The mean of z should be about 0, the variance about 1, and
   the skewness about `population skewness / sqrt(n)` (2 for Exponential,
   0 for Uniform). The skewness shrinks as n grows, which is the CLT at work.

## 5. `central_limit_theorem`: Central Limit Theorem

**คืออะไร:** แสดงว่าค่าเฉลี่ยของตัวอย่างจากประชากรที่ไม่ใช่ Normal จะมีการแจกแจงใกล้ Normal เมื่อ n มากขึ้น

**ทำงานยังไง:**
1. เลือกประชากร Uniform(0,1) หรือ Exp(1) แล้วสุ่ม n ตัว หาค่าเฉลี่ย ทำซ้ำ `reps` รอบ
2. แปลงแต่ละค่าเฉลี่ยเป็น z = (x̄ − μ) / (σ/√n)
3. วาดฮิสโตแกรมของ z เทียบกับจำนวนที่คาดไว้จาก N(0,1) และพิมพ์ mean, var, skewness
ลอง `./central_limit_theorem exp 1`, `exp 5`, `exp 30`, `exp 200` จะเห็นความเบ้ค่อยๆ หายไป

| ส่วนของโปรแกรม | อ้างอิงสไลด์ |
|---|---|
| ทฤษฎีบทขีดจำกัดกลาง: ΣXᵢ ~ N(nμ, nσ²) และ (ΣXᵢ − nμ)/(σ√n) ~ N(0,1) | MGF 15 |
| ตัวอย่างเดียวกันกับ lab: Yᵢ ~ Ex(1) ที่ n = 1, 2, 3, 5, 10 | MGF 16 |
| ผลรวมแล้ว "เป็น Gaussian มากขึ้น" | CM 10 |
| ค่าเฉลี่ยบวกกัน ความแปรปรวนบวกกัน (ตัวแปรอิสระ) | DD 12, CM 8 |
| Uniform(0,1) มี mean 1/2, var 1/12 | CM 8 |
| Exp(1) มี mean 1/λ, var 1/λ² | CD 8 |
| สุ่ม Exp ด้วย `−ln(u)` | CM 15 |
| แปลงเป็นค่ามาตรฐาน (x − μ)/σ | CD 9, CM 14 |
| ความน่าจะเป็นต่อช่วงใช้ Φ (ไม่มีรูปปิด) | CD 10-11 |
| skewness = E[(X−μ)³]/σ³, บวก = เบ้ขวา | CD 14-15 |

⚠ **นอกสไลด์:** ค่าทฤษฎีของ skewness (Exp(1) มี skewness = 2 และ skewness ของค่าเฉลี่ยลดลงตาม 2/√n)

## Source code

One cell per file. Run them all to recreate the project.

````python
%%writefile central_limit_theorem/central_limit_theorem.h
#ifndef CENTRAL_LIMIT_THEOREM_H
# define CENTRAL_LIMIT_THEOREM_H

# include <math.h>
# include <stdio.h>
# include <stdlib.h>
# include <string.h>
# include <time.h>

# define B 16
# define W 40

typedef struct s_clt
{
	int				ex;
	long			n;
	long			reps;
	long			h[B];
	double			mu;
	double			sd;
	double			s1;
	double			s2;
	double			s3;
	unsigned int	seed;
}	t_clt;

double	phi(double z);
void	simulate(t_clt *c);
void	print_hist(t_clt *c);
void	print_stats(t_clt *c);

#endif
````

````python
%%writefile central_limit_theorem/central_limit_theorem.c
#include "central_limit_theorem.h"

static int	parse(t_clt *c, int ac, char **av)
{
	c->ex = ac > 1 && !strcmp(av[1], "exp");
	c->n = 30;
	c->reps = 10000;
	c->seed = (unsigned int)time(NULL);
	if (ac > 2)
		c->n = atol(av[2]);
	if (ac > 3)
		c->reps = atol(av[3]);
	if (ac > 4)
		c->seed = (unsigned int)atol(av[4]);
	if ((ac > 1 && !c->ex && strcmp(av[1], "uniform")) || c->n < 1
		|| c->reps < 1)
	{
		fprintf(stderr, "usage: %s [uniform|exp] [n>=1] [reps>=1] [seed]\n",
			av[0]);
		return (1);
	}
	return (0);
}

static const char	*pop_name(t_clt *c)
{
	if (c->ex)
		return ("Exp(1)");
	return ("Uniform(0,1)");
}

static void	print_head(t_clt *c)
{
	printf("== CLT: population=%s, ", pop_name(c));
	printf("sample size n=%ld, repetitions=%ld, ", c->n, c->reps);
	printf("seed=%u ==\n", c->seed);
	printf("Histogram of z = (xbar - mu)/(sigma/sqrt n): ");
	printf("observed (#) vs Normal(0,1) expected\n");
	printf("%11s %7s %9s\n", "z range", "obs", "expected");
}

int	main(int ac, char **av)
{
	t_clt	c;

	if (parse(&c, ac, av))
		return (1);
	srand(c.seed);
	c.mu = 0.5;
	c.sd = sqrt(1.0 / 12);
	if (c.ex)
	{
		c.mu = 1;
		c.sd = 1;
	}
	memset(c.h, 0, sizeof(c.h));
	simulate(&c);
	print_head(&c);
	print_hist(&c);
	print_stats(&c);
	return (0);
}
````

````python
%%writefile central_limit_theorem/central_limit_theorem_sim.c
#include "central_limit_theorem.h"

double	phi(double z)
{
	return (0.5 * erfc(-z / M_SQRT2));
}

static double	u(void)
{
	return ((rand() + 1.0) / (RAND_MAX + 2.0));
}

static double	zscore(t_clt *c)
{
	long	i;
	double	m;

	m = 0;
	i = -1;
	while (++i < c->n)
	{
		if (c->ex)
			m += -log(u());
		else
			m += u();
	}
	return ((m / c->n - c->mu) / (c->sd / sqrt((double)c->n)));
}

void	simulate(t_clt *c)
{
	long	r;
	double	z;
	int		b;

	c->s1 = 0;
	c->s2 = 0;
	c->s3 = 0;
	r = -1;
	while (++r < c->reps)
	{
		z = zscore(c);
		c->s1 += z;
		c->s2 += z * z;
		c->s3 += z * z * z;
		b = (int)floor((z + 4) * 2);
		if (b >= 0 && b < B)
			c->h[b]++;
	}
}
````

````python
%%writefile central_limit_theorem/central_limit_theorem_print.c
#include "central_limit_theorem.h"

static long	max_count(t_clt *c)
{
	long	mx;
	int		i;

	mx = 1;
	i = -1;
	while (++i < B)
		if (c->h[i] > mx)
			mx = c->h[i];
	return (mx);
}

static void	bar(long v, long mx)
{
	int	k;

	k = 0;
	while (k < (int)(W * (double)v / mx + 0.5))
	{
		putchar('#');
		k++;
	}
	putchar('\n');
}

void	print_hist(t_clt *c)
{
	int		i;
	long	mx;
	double	lo;
	double	e;

	mx = max_count(c);
	i = -1;
	while (++i < B)
	{
		lo = -4 + 0.5 * i;
		e = c->reps * (phi(lo + 0.5) - phi(lo));
		printf("[%+.1f,%+.1f) %7ld %9.1f ", lo, lo + 0.5, c->h[i], e);
		bar(c->h[i], mx);
	}
}

void	print_stats(t_clt *c)
{
	double	m;
	double	v;
	double	sk;
	double	pop;

	m = c->s1 / c->reps;
	v = c->s2 / c->reps - m * m;
	sk = (c->s3 / c->reps - 3 * m * v - m * m * m) / pow(v, 1.5);
	pop = 0;
	if (c->ex)
		pop = 2;
	printf("\nmean(z)=%+.4f (theory 0)  var(z)=%.4f (theory 1)  ", m, v);
	printf("skewness=%+.4f ", sk);
	printf("(theory %+.4f = pop. skew/sqrt n -> 0)\n", pop
		/ sqrt((double)c->n));
	printf("Tip: try n=1, 5, 30, 200 with exp to watch the skew vanish.\n");
}
````

````python
%%writefile central_limit_theorem/Makefile
NAME = central_limit_theorem
SRCS = central_limit_theorem.c central_limit_theorem_sim.c central_limit_theorem_print.c
CFLAGS = -Wall -Wextra -O2
ARGS = exp 30 10000 1

all: $(NAME)

$(NAME): $(SRCS) central_limit_theorem.h
	$(CC) $(CFLAGS) $(SRCS) -o $(NAME) -lm

# make run               runs the example above
# make run ARGS="..."    runs with your own arguments
run: $(NAME)
	./$(NAME) $(ARGS)

clean:
	rm -f $(NAME)

re: clean all

.PHONY: all run clean re
````
